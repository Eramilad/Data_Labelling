```python
##define_the_labeling_function

def amount_spent(df):
    total = df['overall_gross'].sum()
    return total
```

```python
##LabelMaker to run a search to automatically extract training examples from historical data. 

label_maker = cp.LabelMaker(
    target_dataframe_index = 'top_release',
    time_index = 'date',
    labeling_function = amount_spent,
    window_size = '28d',
)
```

```python
%pip install --upgrade composeml
```

```python
labels = label_maker.search(
    df.sort_values('date'),
    num_examples_per_instance =-1,
    gap =1,
    verbose = True,
)
labels.head()
```

```python
df['start_date'] = pd.to_datetime(
    df['date'].str.extract(r'^([A-Za-z]{3} \d{1,2})')[0] + ' 2026',
    format='%b %d %Y',
    errors='coerce'
)
```

```python
print(df[['date', 'start_date']].to_string())
```

```python
label_maker = cp.LabelMaker(
    target_dataframe_index='top_release',
    time_index='start_date',
    labeling_function=amount_spent,
    window_size='28d',
)
```

```python
labels = label_maker.search(
    df.sort_values('start_date'),
    num_examples_per_instance=-1,
    gap=1,
    verbose=True,
)

labels.head()
```
