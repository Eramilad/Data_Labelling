```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

img1 = Image.open(r'C:\Users\Damil\Downloads\White Flower.jpg')

img2 = Image.open(r'C:\Users\Damil\Downloads\White Flower.jpg')
img1_array = np.array(img1)

img2_array = np.array(img2)

fig, axes = plt.subplots(nrows=1, ncols=2)

axes[0].imshow(img1_array)

axes[1].imshow(img2_array)

plt.show()
```

```python
img = Image.open(r'C:\Users\Damil\Downloads\White Flower.jpg')
r, g, b  = img.split()
```

```python
r_array = np.array(r)
g_array = np.array(g)
b_array = np.array(b)
```

```python
fig, axes = plt.subplots(nrows=1, ncols=3)

axes[0].imshow(r_array, cmap='Reds')

axes[1].imshow(g_array, cmap='Greens')

axes[2].imshow(b_array, cmap='Blues')

plt.show()
```
