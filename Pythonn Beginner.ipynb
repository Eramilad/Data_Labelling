```python
# Import statements.
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import seaborn as sns
```

```python
x = [10, 'gold', 'dollars']
```

```python
def f1_score(precision,recall):
    score = 2*precision*recall / (precision + recall)
    score = np.nan_to_num(score)

    return score
```

```python
print (f1_score(0.2,0.4))
```

```python
x = np.linspace(0,10,1001)
y = np.linspace(0,10,1001)
X,Y = np.meshgrid(x, y)
Z = f1_score(X,Y)
ab = plt.figure()
ab.set_size_inches(10,10)
ax = plt.axes(projection= '3d')
ax.plot_surface(X,Y,Z, cstride = 2, rstride = 3, cmap = 'plasma')

ax.set_title('$F_{10}$ of precision,recall', size = 18)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.view_init(35, -65)
```
